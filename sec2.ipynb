```python
#Ques -- 2
num1 = int(input("Enter the number:"))
num2 = int(input("Enter the number:"))
operator = input("Enter the operator:(+,-,*,/,%,//,**)")
if operator == "+":
    print("The sum of the num1 and num2 is:",num1+num2)
elif operator == "-":
    print("The subtraction of the num1 and num2 is:",num1-num2)
elif operator == "*":
    print("The multiplication of the num1 and num2 is:",num1*num2)
elif operator == "/":
    print("The division of the num1 and num2 is:",num1/num2)
elif operator == "%":
    print("The modulo of the num1 and num2 is:",num1%num2)
elif operator == "//":
    print("The floor of the num1 and num2 is:",num1//num2)
elif operator == "**":
     print("The exponention of the num1 and num2 is:",num1**num2)
```
