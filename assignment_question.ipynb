**Question 17:** Write a Python program to check if a year is a leap year.

```python
# Your code here
year=int(input("Enter the number:"))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print("The year is a leap year:",year)
        else:
            print("The year is not a leap year:",year)
    else:
           print("The year is  a leap year:",year)
else:
        print("The year is not a leap year:",year)
```
