```python
def password(P):
    if len(P) <= 5:
        return False
    if not any(char.isdigit() for char in P):
        return False
    if not any(char.islower()for char in P):
        return False
    if not any(char.upper()for char in P):
        return False
    if not any(char in '@#$%*!' for char in P):
        return False
    return True
print(password("cfEfyg34ug@#$%*!"))
```

```python
# Input 3 numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

# HCF / GCD
small = min(a, b, c)

for i in range(1, small + 1):
    if a % i == 0 and b % i == 0 and c % i == 0:
        hcf = i

# LCM
lcm = max(a, b, c)

while lcm:
    if lcm % a == 0 and lcm % b == 0 and lcm % c == 0:
        break
    lcm = lcm + 1

# Output
print("HCF =", hcf)
print("GCD =", hcf)
print("LCM =", lcm)
```
