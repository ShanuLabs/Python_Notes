```python
count=0
lst1 = [1, 2, 3]
lst2 = [1, 2, 3, 4, 5]
for i in lst2:
    for j in lst1:
        if i==j:
            count=count+1
if count==len(lst1):
    print("yes it is a subset")
else:
    print("Not")
```
