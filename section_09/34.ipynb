```python
##Redading a filew therough a prg
with open("sample.txt","r") as file:
    reader=file.read()
    print(reader)
```

```python
##Reading file line by line
with open("sample.txt","r") as file:
    for line in file:
        print(line.strip())
```

```python
##Overwrittng a file
with open("sample.txt","w") as file:
    file.write("Hello world")
with open("sample.txt","r") as file:
    reader=file.read()
    print(reader)
```

```python
with open("sample.txt","a") as file:
    file.write("Congrats for getting a job")
with open("sample.txt","r") as file:
    reader=file.read()
    print(reader.strip())
    
```

```python
file = open("sample.txt", "r")

text = file.read()

lines = text.splitlines()
words = text.split()
characters = len(text)

print("Lines:", len(lines))
print("Words:", len(words))
print("Characters:", characters)

file.close()
```

```python
l = ["Riya\t","i\t","really\t","love\t","you\t"]
with open("sample.txt","w+") as file:
    file.writelines(l)
    file.seek(0)
    print(file.read())
```
