```python
memo = {}

def fib(num):

    if num <= 0:
        return 0

    elif num == 1:
        return 1

    elif num in memo:
        return memo[num]

    else:
        memo[num] = fib(num-1) + fib(num-2)
        return memo[num]

num = int(input("Enter the number: "))

print(fib(num))
```

```python
def separate(lst):

    integers = []
    strings = []
    floats = []

    for i in lst:

        if type(i) == int:
            integers.append(i)

        elif type(i) == str:
            strings.append(i)

        elif type(i) == float:
            floats.append(i)

    return integers, strings, floats


lst = [10, "Shanu", 20.5, 30, "Python", 50.7]

a, b, c = separate(lst)

print("Integers :", a)
print("Strings :", b)
print("Floats :", c)
```
